#1. Explain the difference between a Tableau Dashboard and a Tableau Story Point. Give two healthcare-related use cases.

**Answer:** A Tableau Dashboard is a collection of multiple visualizations displayed on a single screen. It helps users analyze different aspects of data together.

A Tableau Story Point is a sequence of visualizations that presents data in a meaningful order to explain insights and support decision-making.
Tableau Dashboard	Tableau Story Point
Combines multiple charts in one view	Presents information in a sequence
Used for interactive analysis	Used for storytelling
Allows users to explore data	Guides users through insights
Example: Hospital performance dashboard	Example: Story showing patient trends and treatment costs

Two healthcare use cases:
1. A hospital dashboard can display patient distribution, department costs, and patient outcomes.
2. A Tableau Story can explain how patient admissions, treatment costs, and outcomes change over time.

#2. What are Actions in Tableau? Explain how filter action can hospital management analyze patient data.

**Answer:** Actions in Tableau allow users to interact with visualizations. They help make dashboards dynamic and interactive.

Types of Tableau Actions:
- Filter Action
- Highlight Action
- URL Action
- Go to Sheet Action
- Parameter Action
Filter Action in Hospital Management
A Filter Action allows users to select a particular region, department, or chart element and filter other visualizations accordingly.

Example:

If hospital management clicks on the West region in the Region Chart, the Department Cost Chart and Outcome Chart will display information related to the West region only.



Insight: Filter Actions help hospital management analyze regional patient trends, department expenses, and patient outcomes without creating separate dashboards.
